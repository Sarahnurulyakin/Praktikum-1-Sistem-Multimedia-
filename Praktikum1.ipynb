```python
### Eksplorasi Dasar Dengan Python 
#### Membaca dan Menampilkan Gambar 

import cv2 
from matplotlib import pyplot as plt 
# Membaca gambar 
image = cv2.imread('praktikum1.jpg')  # Ganti 'sample.jpg' dengan path gambar Anda 
# Konversi warna dari BGR ke RGB 
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB) 
# Menampilkan gambar 
plt.imshow(image_rgb) 
plt.title('Gambar Awal') 
plt.axis('off') 
plt.show() 
```

```python
#### Tugas Ganti dalam warna Greyscale

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
plt.imshow(gray, cmap='gray')
plt.show()
```

```python
#### Operasi Dasar Pada Gambar

# Resize gambar 
resized_image = cv2.resize(image, (200, 200))  # Resize ke ukuran 200x200 
# Crop gambar 
cropped_image = image[50:150, 50:150]  # Crop bagian tertentu 
# Menampilkan hasil manipulasi 
plt.subplot(1, 2, 1) 
plt.imshow(cv2.cvtColor(resized_image, cv2.COLOR_BGR2RGB)) 
plt.title('Resize') 
plt.subplot(1, 2, 2) 
plt.imshow(cv2.cvtColor(cropped_image, cv2.COLOR_BGR2RGB)) 
plt.title('Crop') 
plt.show() 
```

```python
#### Tugas Ubah Ukuran dan Crop Bagian Lain

# Resize gambar
resized_image = cv2.resize(image, (300, 300))  # ukuran 300x300

# Crop bagian berbeda ( kanan bawah)
cropped_image = image[100:200, 100:200]

# Menampilkan hasil
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(resized_image, cv2.COLOR_BGR2RGB))
plt.title('Resize')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(cropped_image, cv2.COLOR_BGR2RGB))
plt.title('Crop')

plt.show()
```

```python
#### Membuat Histogram Gambar

colors = ('b', 'g', 'r')  # Warna channel 
for i, color in enumerate(colors): histogram = cv2.calcHist([image], [i], None, [256], [0, 
256]) 
plt.plot(histogram, color=color) 
plt.title('Histogram Warna') 
plt.xlabel('Intensitas') 
plt.ylabel('Jumlah Piksel') 
plt.show() 
```

```python
#### Tugas: Gunakan gambar grayscale dan bandingkan histogramnya dengan gambar berwarna.

# Ubah ke grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# =========================
# Histogram Warna
# =========================
colors = ('b', 'g', 'r')
for i, color in enumerate(colors):
    histogram = cv2.calcHist([image], [i], None, [256], [0,256])
    plt.plot(histogram, color=color)

plt.title('Histogram Warna')
plt.xlabel('Intensitas')
plt.ylabel('Jumlah Piksel')
plt.show()

# =========================
# Histogram Grayscale
# =========================
hist_gray = cv2.calcHist([gray], [0], None, [256], [0,256])

plt.plot(hist_gray, color='black')
plt.title('Histogram Grayscale')
plt.xlabel('Intensitas')
plt.ylabel('Jumlah Piksel')
plt.show()
```

## Tugas Mandiri 


```python
### Baca Gambar Dan Resize Ukuran Gambar
 
import cv2
import matplotlib.pyplot as plt

# Baca gambar baru
image = cv2.imread('Tugas Mandiri.jpg')

# Resize
resize = cv2.resize(image, (300, 300))

# Tampilkan gambar asli & resize
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Gambar Asli')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(resize, cv2.COLOR_BGR2RGB))
plt.title('Resize 300x300')
plt.axis('off')

plt.show()
```

```python
#### Crop Bagian Tengah Gambar 

h, w = image.shape[:2]
crop = image[h//4:h*3//4, w//4:w*3//4]

plt.imshow(cv2.cvtColor(crop, cv2.COLOR_BGR2RGB))
plt.title('Crop Tengah')
plt.axis('off')
plt.show()
```

```python
#### Menampilkan Histogram Greyscale

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

hist = cv2.calcHist([gray], [0], None, [256], [0,256])

plt.plot(hist, color='black')
plt.title('Histogram Grayscale')
plt.xlabel('Intensitas')
plt.ylabel('Jumlah Piksel')
plt.show()
```
